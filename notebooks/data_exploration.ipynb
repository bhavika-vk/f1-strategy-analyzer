```python
pit_stops = laps[laps['PitInTime'].notna()][['Driver', 'LapNumber']]
```

```python
from itertools import groupby
```

```python
summary = laps.groupby('Driver').agg(
    num_pitstops=('PitInTime', lambda x: x.notna().sum()),
    compounds_used=('Compound', lambda x: [k for k, _ in groupby(x)]),
).reset_index()
```

```python
pit_laps_by_driver = pit_stops.groupby('Driver')['LapNumber'].apply(list).reset_index()
pit_laps_by_driver.columns = ['Driver', 'pit_stop_laps']
```

```python
summary = summary.merge(pit_laps_by_driver, on='Driver', how='left')
```

```python
summary
```

```python
pit_counts = laps.groupby('Driver')['PitInTime'].apply(lambda x: x.notna().sum())
multi_pit_drivers = pit_counts[pit_counts >= 2].index
```

```python
filtered = laps[(laps['Driver'].isin(multi_pit_drivers)) & (laps['LapNumber'] != 1) & (laps['PitInTime'].isna())]
```

```python
rows, cols = 2, 3
fig, axes = plt.subplots(rows, cols, figsize=(15, 8), sharex=True, sharey=True)
axes = axes.flatten()

linestyles = {'SOFT': 'solid', 'MEDIUM': 'dashed', 'HARD': 'dotted'}

for i, drv in enumerate(multi_pit_drivers):
    drv_laps = filtered[filtered['Driver'] == drv]
    for compound in drv_laps['Compound'].unique():
        comp_laps = drv_laps[drv_laps['Compound'] == compound].sort_values('TyreLife')
        axes[i].plot(comp_laps['TyreLife'], comp_laps['LapTime'].dt.total_seconds(),
                     linestyle=linestyles.get(compound, 'solid'),
                     marker='o', markersize=3, label=compound)
    axes[i].set_title(drv)
    axes[i].legend(fontsize=8)

fig.suptitle('Lap Time vs Tyre Age by Compound (Monaco 2024)', fontsize=14)
fig.text(0.5, 0.04, 'Tyre Life (laps)', ha='center')
fig.text(0.08, 0.5, 'Lap Time (s)', va='center', rotation='vertical')
plt.tight_layout(rect=[0.08, 0.05, 1, 0.96])
plt.show()
```

```python
warmup_check = laps[(laps['TyreLife'].isin([1, 2, 3, 4, 5])) & (laps['LapTime'].dt.total_seconds() < 200)]
```

```python
linestyles = {'SOFT': 'solid', 'MEDIUM': 'dashed', 'HARD': 'dotted'}

fig, ax = plt.subplots(figsize=(12, 7))

for (drv, stint), group in warmup_check.groupby(['Driver', 'Stint']):
    group = group.sort_values('TyreLife')
    compound = group['Compound'].iloc[0]
    ax.plot(group['TyreLife'], group['LapTime'].dt.total_seconds(), marker='o', alpha=0.6, linewidth=1)
    first_x = group['TyreLife'].iloc[0]
    first_y = group['LapTime'].dt.total_seconds().iloc[0]
    ax.text(first_x - 0.15, first_y, f'{drv} ({compound})', fontsize=7, va='center', ha='right')

ax.set_xlabel('Tyre Life (laps)')
ax.set_ylabel('Lap Time (s)')
ax.set_title('Lap Time by Tyre Age (1-5) — Every Driver/Stint (Monaco 2024)')
plt.tight_layout()
plt.show()
```
