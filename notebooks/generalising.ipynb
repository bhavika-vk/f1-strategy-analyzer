```python
# TRACK CATEGORIZATION APPROACH
# Categorize tracks by track-level properties (not one race's events), so the model
# can be trained on 1-2 tracks per category and tested on the others in that category.
#
# From FastF1 - fastest dry lap of qualifying (fallback: sprint qualifying, then FP3):
# 1. Avg lap speed      -> primary layout signal (flowing vs technical)
# 2. Top speed          -> isolates straight-line length (Baku: high top speed,
#                          moderate avg speed due to the technical sector)
# 3. Full-throttle %    -> share of the lap at full throttle (Monza ~80%, Monaco ~45%)
# 4. Number of turns    -> layout complexity, independent of speed
#
# From FastF1 races:
# 5. Median pit stops per driver, averaged over 2022-2024 -> degradation proxy;
#                          averaging smooths one-off events (red flags, safety cars)
#
# From Pirelli 2024 previews:
# 6. Ratings (1-5): traction, tyre stress, asphalt grip, braking, asphalt abrasion,
#                   lateral, track evolution, downforce
# 7. Hardest compound (C1-C5) -> lower = harder compound = higher tyre stress
# 8. Circuit length     -> context only
#
# Dropped: TrackStatus share (race-specific), laps and race distance (redundant
# with length), medium/soft compounds (always hard + 1/+2).
#
# Workflow: build the joined table -> normalize -> check collinearity -> cluster
# (hierarchical, 3-4 categories, >= 3 tracks each).
```

```python
import pandas as pd
import numpy as np
```

**Pirelli 2024 track ratings**

Source: https://www.pirelli.com/global/en-ww/emotions-and-numbers/infographics-2024/

```python
pirelli = pd.read_csv('../data/pirelli_2024_track_ratings.csv')
print(pirelli.shape, '| missing:', pirelli.isna().sum().sum(), '| duplicates:', pirelli.duplicated().sum())
print(pirelli.dtypes)
pirelli.head()
```

```python
RATING_COLS = ['traction', 'tyre_stress', 'asphalt_grip', 'braking',
               'asphalt_abrasion', 'lateral', 'track_evolution', 'downforce']

# medium/soft are always hard + 1/+2; laps and race_distance_km are redundant with length
features = (
    pirelli
    .assign(hardest_compound=pirelli['hard'].str[1].astype(int))
    .set_index('race')[RATING_COLS + ['hardest_compound', 'circuit_length_km']]
)
features_scaled = (features - features.mean()) / features.std()
print(features_scaled.round(2))
```

```python
HIGH_CORR = 0.7

corr = features.corr()
print(corr.round(2))
pairs = corr.where(np.triu(np.ones(corr.shape, dtype=bool), k=1)).stack()
print(pairs[pairs.abs() >= HIGH_CORR].sort_values())
```

```python
for col in features.columns:
    ranked = features[col].sort_values()
    print(f'{col}: low {list(ranked.index[:3])} | high {list(ranked.index[-3:])}')
```

```python
QUALI_SESSIONS = ['Qualifying', 'Sprint Qualifying', 'Practice 3', 'Practice 2', 'Practice 1']  # fallback order if wet
DRY_COMPOUNDS = {'SOFT', 'MEDIUM', 'HARD'}
FULL_THROTTLE_MIN = 95   # fixed, not relative to the lap max, so sensor spikes (e.g. 104) don't shift it
EXPECTED_MAX_THROTTLE = 100
PLAUSIBLE_THROTTLE_PCT = (30, 90)   # 2024 range is ~45 (Monaco) to ~76 (Monza)


def get_track_metrics(year, round_number):
    """Layout metrics from the fastest dry lap of the first dry qualifying-type session."""
    event = fastf1.get_event(year, round_number)
    for session_name in QUALI_SESSIONS:
        try:
            session = event.get_session(session_name)
        except ValueError:  # session not held this weekend
            continue
        session.load(weather=False, messages=False)
        fastest_lap = session.laps.pick_fastest()
        if fastest_lap is not None and fastest_lap['Compound'] in DRY_COMPOUNDS:
            break
    else:
        raise ValueError(f'No dry session found for {year} round {round_number}')

    tel = fastest_lap.get_car_data()
    max_throttle = tel['Throttle'].max()
    full_throttle_pct = (tel['Throttle'] > FULL_THROTTLE_MIN).mean() * 100

    label = f'{year} round {round_number} {session_name}'
    if max_throttle != EXPECTED_MAX_THROTTLE:
        print(f'GLITCH? {label}: throttle peaks at {max_throttle}, expected {EXPECTED_MAX_THROTTLE}')
    if not PLAUSIBLE_THROTTLE_PCT[0] <= full_throttle_pct <= PLAUSIBLE_THROTTLE_PCT[1]:
        print(f'GLITCH? {label}: full throttle {full_throttle_pct:.1f}% outside {PLAUSIBLE_THROTTLE_PCT}')

    return {
        'round': round_number,
        'source_session': session_name,
        'avg_speed': tel['Speed'].mean(),
        'top_speed': tel['Speed'].max(),
        'full_throttle_pct': full_throttle_pct,
        'max_throttle': max_throttle,
        'num_turns': len(session.get_circuit_info().corners),
    }
```

```python
PIT_YEARS = [2022, 2023, 2024]
schedules = {year: fastf1.get_event_schedule(year, include_testing=False) for year in PIT_YEARS}


def get_median_pitstops(event_name):
    """Median pit stops per driver, averaged over the PIT_YEARS the race was held."""
    yearly_medians = []
    for year, schedule in schedules.items():
        rounds = schedule.loc[schedule['EventName'] == event_name, 'RoundNumber']
        if rounds.empty:  # race not on that year's calendar
            continue
        session = fastf1.get_session(year, int(rounds.iloc[0]), 'R')
        session.load(telemetry=False, weather=False, messages=False)
        if session.laps.empty:  # e.g. cancelled race
            continue
        stops_per_driver = session.laps.groupby('Driver')['PitInTime'].count()
        yearly_medians.append(stops_per_driver.median())
    return np.mean(yearly_medians)
```

```python
schedule_2024 = fastf1.get_event_schedule(2024, include_testing=False)

track_rows = []
for _, event in schedule_2024.iterrows():
    try:
        metrics = get_track_metrics(2024, event['RoundNumber'])
        metrics['median_pitstops'] = get_median_pitstops(event['EventName'])
        track_rows.append(metrics)
    except Exception as err:  # keep looping; failed rounds are listed and re-run individually
        print(f"Round {event['RoundNumber']} ({event['EventName']}) failed: {err}")

track_metrics = pd.DataFrame(track_rows)
track_table = pirelli.merge(track_metrics, on='round', how='left')
print(track_table.shape, '| rounds missing FastF1 metrics:', track_table['avg_speed'].isna().sum())
track_table.head()
```

```python
FASTF1_COLS = ['avg_speed', 'top_speed', 'full_throttle_pct', 'num_turns', 'median_pitstops']

features = (
    track_table
    .assign(hardest_compound=track_table['hard'].str[1].astype(int))
    .set_index('race')[FASTF1_COLS + RATING_COLS + ['hardest_compound', 'circuit_length_km']]
)
with pd.option_context('display.max_rows', None, 'display.max_columns', None, 'display.width', None):
    display(features.round(2))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots(figsize=(12, 10))
sns.heatmap(features.corr(), annot=True, fmt='.2f', cmap='RdBu_r', vmin=-1, vmax=1,
            square=True, cbar_kws={'label': 'Pearson r'}, ax=ax)
ax.set_title('Feature correlation across 2024 tracks')
plt.tight_layout()
plt.show()
```

```python
# Keep one feature per highly correlated group so it isn't double-weighted in distances:
# full_throttle_pct ~ avg_speed (r 0.95); lateral, hardest_compound ~ tyre_stress (|r| 0.74-0.84)
DROP_COLS = ['full_throttle_pct', 'lateral', 'hardest_compound']

cluster_features = features.drop(columns=DROP_COLS)
cluster_scaled = (cluster_features - cluster_features.mean()) / cluster_features.std()
print(cluster_scaled.shape, '|', list(cluster_scaled.columns))
```

```python
from scipy.cluster.hierarchy import dendrogram, fcluster, linkage
from sklearn.metrics import silhouette_score

linkage_matrix = linkage(cluster_scaled, method='ward')

fig, ax = plt.subplots(figsize=(12, 6))
dendrogram(linkage_matrix, labels=cluster_scaled.index.tolist(), leaf_rotation=90, ax=ax)
ax.set_title('Ward hierarchical clustering of 2024 tracks (z-scored features)')
ax.set_ylabel('Ward linkage distance')
plt.tight_layout()
plt.show()
```

```python
CLUSTER_COUNTS = [3, 4]
MIN_TRACKS_PER_CLUSTER = 3

cluster_labels = pd.DataFrame(
    {k: fcluster(linkage_matrix, t=k, criterion='maxclust') for k in CLUSTER_COUNTS},
    index=cluster_scaled.index,
)
for k in CLUSTER_COUNTS:
    sizes = cluster_labels[k].value_counts().sort_index()
    silhouette = silhouette_score(cluster_scaled, cluster_labels[k])
    print(f'k={k}: sizes {sizes.tolist()} | all >= {MIN_TRACKS_PER_CLUSTER}: '
          f'{(sizes >= MIN_TRACKS_PER_CLUSTER).all()} | silhouette {silhouette:.2f}')
    for cluster_id, tracks in cluster_labels.groupby(k).groups.items():
        print(f'  {cluster_id}: {list(tracks)}')
    print('Mean z-score per cluster:')
    print(cluster_scaled.groupby(cluster_labels[k]).mean().round(2).T, '\n')
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score

RANDOM_SEED = 42
KMEANS_N_INIT = 50   # many restarts; cheap on 24 rows and avoids a poor local optimum

for k in CLUSTER_COUNTS:
    kmeans_labels = KMeans(n_clusters=k, n_init=KMEANS_N_INIT, random_state=RANDOM_SEED).fit_predict(cluster_scaled)
    ari = adjusted_rand_score(cluster_labels[k], kmeans_labels)
    silhouette = silhouette_score(cluster_scaled, kmeans_labels)
    print(f'k={k}: ARI vs Ward {ari:.2f} | k-means silhouette {silhouette:.2f}')
    print(pd.crosstab(cluster_labels[k], kmeans_labels, rownames=['ward'], colnames=['kmeans']))
    for cluster_id, tracks in cluster_scaled.groupby(kmeans_labels).groups.items():
        print(f'  kmeans {cluster_id}: {list(tracks)}')
    print()
```

```python
MAX_CLUSTERS = 8

rows = []
for k in range(2, MAX_CLUSTERS + 1):
    ward_labels = fcluster(linkage_matrix, t=k, criterion='maxclust')
    kmeans_labels = KMeans(n_clusters=k, n_init=KMEANS_N_INIT, random_state=RANDOM_SEED).fit_predict(cluster_scaled)
    rows.append({
        'k': k,
        'ward_silhouette': silhouette_score(cluster_scaled, ward_labels),
        'kmeans_silhouette': silhouette_score(cluster_scaled, kmeans_labels),
        'ward_min_size': pd.Series(ward_labels).value_counts().min(),
        'kmeans_min_size': pd.Series(kmeans_labels).value_counts().min(),
        'ari_ward_vs_kmeans': adjusted_rand_score(ward_labels, kmeans_labels),
    })
k_comparison = pd.DataFrame(rows).set_index('k')
print(k_comparison.round(2))

# (column prefix, legend label, label offset in points) - offsets keep the two methods' labels apart
methods = [('ward', 'Ward (hierarchical)', 8), ('kmeans', 'K-means', -14)]

fig, ax = plt.subplots(figsize=(8, 5))
for (prefix, label, offset), color in zip(methods, sns.color_palette('colorblind')):
    scores = k_comparison[f'{prefix}_silhouette']
    ax.plot(scores.index, scores, marker='o', linewidth=2, color=color, label=label)
    for k, score in scores.items():
        ax.annotate(f"min {k_comparison.loc[k, f'{prefix}_min_size']}", (k, score),
                    textcoords='offset points', xytext=(0, offset), ha='center', fontsize=8)
ax.set_title('Cluster separation vs number of clusters\n(point labels: smallest cluster size)')
ax.set_xlabel('Number of clusters (k)')
ax.set_ylabel('Mean silhouette score')
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Smaller, degradation-focused feature set: 12 features on 24 tracks gave weak structure (silhouette <= 0.19)
DEG_FEATURE_COLS = ['tyre_stress', 'asphalt_abrasion', 'avg_speed', 'downforce', 'median_pitstops']

deg_features = features[DEG_FEATURE_COLS]
deg_scaled = (deg_features - deg_features.mean()) / deg_features.std()
deg_linkage = linkage(deg_scaled, method='ward')

rows = []
for k in range(2, MAX_CLUSTERS + 1):
    ward_labels = fcluster(deg_linkage, t=k, criterion='maxclust')
    kmeans_labels = KMeans(n_clusters=k, n_init=KMEANS_N_INIT, random_state=RANDOM_SEED).fit_predict(deg_scaled)
    rows.append({
        'k': k,
        'ward_silhouette': silhouette_score(deg_scaled, ward_labels),
        'kmeans_silhouette': silhouette_score(deg_scaled, kmeans_labels),
        'ward_min_size': pd.Series(ward_labels).value_counts().min(),
        'kmeans_min_size': pd.Series(kmeans_labels).value_counts().min(),
        'ari_ward_vs_kmeans': adjusted_rand_score(ward_labels, kmeans_labels),
    })
deg_k_comparison = pd.DataFrame(rows).set_index('k')
print(deg_k_comparison.round(2))

fig, ax = plt.subplots(figsize=(8, 5))
for (prefix, label, offset), color in zip(methods, sns.color_palette('colorblind')):
    scores = deg_k_comparison[f'{prefix}_silhouette']
    ax.plot(scores.index, scores, marker='o', linewidth=2, color=color, label=label)
    for k, score in scores.items():
        ax.annotate(f"min {deg_k_comparison.loc[k, f'{prefix}_min_size']}", (k, score),
                    textcoords='offset points', xytext=(0, offset), ha='center', fontsize=8)
ax.set_title('Cluster separation vs number of clusters, 5 degradation features\n(point labels: smallest cluster size)')
ax.set_xlabel('Number of clusters (k)')
ax.set_ylabel('Mean silhouette score')
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
DEG_CLUSTER_COUNTS = [3, 4]   # k=3 best silhouette, k=4 best Ward/k-means agreement

for k in DEG_CLUSTER_COUNTS:
    ward_labels = pd.Series(fcluster(deg_linkage, t=k, criterion='maxclust'), index=deg_scaled.index)
    kmeans_labels = pd.Series(
        KMeans(n_clusters=k, n_init=KMEANS_N_INIT, random_state=RANDOM_SEED).fit_predict(deg_scaled),
        index=deg_scaled.index,
    )
    crosstab = pd.crosstab(ward_labels, kmeans_labels, rownames=['ward'], colnames=['kmeans'])
    print(f'===== k={k} | ARI {adjusted_rand_score(ward_labels, kmeans_labels):.2f}')
    print(crosstab)
    for cluster_id, tracks in ward_labels.groupby(ward_labels).groups.items():
        print(f'  ward {cluster_id} ({len(tracks)}): {list(tracks)}')
    # k-means numbering is arbitrary: map each Ward cluster to the k-means cluster it overlaps most
    best_match = crosstab.idxmax(axis=1)
    print('  k-means disagrees on:', list(ward_labels.index[kmeans_labels != ward_labels.map(best_match)]))
    print('Mean z-score per Ward cluster:')
    print(deg_scaled.groupby(ward_labels).mean().round(2).T, '\n')
```

```python
FINAL_K = 4
CLUSTER_NAMES = {1: 'fast_low_wear', 2: 'slow_high_downforce', 3: 'high_stress_high_deg', 4: 'balanced_abrasive'}
# Borderline track where Ward and k-means disagree; placed by layout (downforce 5, slow), see decision log.
# Imola, the other borderline track, stays in Ward's balanced_abrasive cluster.
MANUAL_ASSIGNMENTS = {'Hungary': 'slow_high_downforce'}

track_category = (
    pd.Series(fcluster(deg_linkage, t=FINAL_K, criterion='maxclust'), index=deg_scaled.index)
    .map(CLUSTER_NAMES)
)
track_category.update(pd.Series(MANUAL_ASSIGNMENTS))
print(track_category.value_counts().to_string(), '\n')
for category, tracks in track_category.groupby(track_category).groups.items():
    print(f'{category}: {list(tracks)}')
```

```python
def get_incident_rate(year, round_number):
    """Percent of race laps not fully green (TrackStatus != '1': yellow, SC, VSC, red flag)."""
    session = fastf1.get_session(year, round_number, 'R')
    session.load(telemetry=False, weather=False, messages=False)
    return (session.laps['TrackStatus'] != '1').mean() * 100


# Race-specific, so not a clustering feature; used to profile groups and pick clean races
incident_rate = pd.Series(
    {race: get_incident_rate(2024, round_number) for race, round_number in zip(pirelli['race'], pirelli['round'])}
)

PROFILE_COLS = ['avg_speed', 'num_turns', 'tyre_stress', 'asphalt_abrasion', 'median_pitstops']
track_profile = features[PROFILE_COLS].assign(incident_rate_pct=incident_rate, track_type=track_category)

display(track_profile.groupby('track_type').mean().style.format('{:.2f}').set_caption('Group averages'))
display(
    track_profile.reset_index()
    .sort_values(['track_type', 'incident_rate_pct'])
    .set_index(['track_type', 'race'])
    .style.format('{:.2f}')
    .set_caption('Tracks by group, cleanest race first')
)
```
